# ResNet50 224 — pipeline end-to-end para detecção de rosto sintético

Este notebook começa no dataset cru (`training/dataset`), aplica YuNet para localizar uma única face, cria crops PNG quadrados com margem de 15%, audita os dados, divide as partições, treina e salva um artefato autocontido.

O código disponibiliza duas estratégias de divisão, mas executa somente a indicada em `SCENARIO`. O experimento documentado utiliza `source_holdout`: `sd_21` (fake) e `celebhq` (real) são fontes inteiras reservadas para teste. A opção `random_image` é uma divisão aleatória alternativa e não representa comparação controlada com outros notebooks.

O teste nunca é usado para escolher época, arquitetura ou limiar. A validação escolhe o checkpoint e o limiar.

Definição dos hiperparâmetros de treinamento, configurações do detector facial YuNet, diretórios de dados e modelos e dispositivo de processamento (CPU/GPU).

## Pré-processamento end-to-end com YuNet

Esta etapa é a única origem dos dados de treino. Cada imagem do dataset cru recebe no máximo um crop: imagens sem face, com múltiplas faces ou com falha de leitura são registradas no manifesto e excluídas de forma explícita.

Os crops são PNG quadrados de 224×224. O cache é reutilizado apenas quando `REBUILD_CROPS=False`; use `True` ao trocar detector, margem, tamanho ou dataset cru.

```python
CLASS_NAMES = ['fake', 'real']
CLASS_TO_INDEX = {name: index for index, name in enumerate(CLASS_NAMES)}
MEAN = [0.485, 0.456, 0.406]
STD = [0.229, 0.224, 0.225]

raw_dataset = ImageFolder(RAW_DATASET_DIR)
assert raw_dataset.classes == CLASS_NAMES, (
    f'Classes inesperadas: {raw_dataset.classes}. Esperadas: {CLASS_NAMES}.'
)

# Detecta colisões de nome antes de escrever crops. Não sobrescreva uma imagem por outra.
output_paths = {}
for source_path, label in raw_dataset.samples:
    source_path = Path(source_path)
    output_path = CROPS_DIR / raw_dataset.classes[label] / f'{source_path.stem}.png'
    if output_path in output_paths:
        raise RuntimeError(
            f'Colisão de nomes: {source_path} e {output_paths[output_path]} gerariam {output_path}. '
            'Renomeie os arquivos de origem antes de continuar.'
        )
    output_paths[output_path] = source_path

face_detector = FaceDetector(score_threshold=FACE_DETECTOR_SCORE_THRESHOLD)
preprocessing_rows = []

for source_path, label in tqdm(raw_dataset.samples, desc='Criando/verificando crops YuNet', unit='imagem'):
    source_path = Path(source_path)
    class_name = raw_dataset.classes[label]
    output_path = CROPS_DIR / class_name / f'{source_path.stem}.png'
    output_path.parent.mkdir(parents=True, exist_ok=True)

    row = {
        'source_path': str(source_path),
        'crop_path': str(output_path),
        'class_name': class_name,
        'label': label,
        'source': source_path.stem.split('_', maxsplit=1)[1],
        'status': 'valid',
        'created_or_rebuilt': False,
    }

    if output_path.is_file() and not REBUILD_CROPS:
        preprocessing_rows.append(row)
        continue

    image = cv2.imread(str(source_path))
    if image is None:
        row['status'] = 'read_error'
        preprocessing_rows.append(row)
        continue

    crop, status = face_detector.crop_face(image, margin=FACE_MARGIN)
    if status != 'valid' or crop is None:
        row['status'] = status
        preprocessing_rows.append(row)
        continue

    crop = cv2.resize(crop, (IMAGE_SIZE, IMAGE_SIZE), interpolation=cv2.INTER_AREA)
    temporary_path = output_path.with_suffix('.tmp.png')
    if not cv2.imwrite(str(temporary_path), crop):
        row['status'] = 'write_error'
        preprocessing_rows.append(row)
        continue
    temporary_path.replace(output_path)
    row['created_or_rebuilt'] = True
    preprocessing_rows.append(row)

preprocessing_manifest = pd.DataFrame(preprocessing_rows)
manifest_path = CROPS_DIR / 'preprocessing_manifest.csv'
preprocessing_manifest.to_csv(manifest_path, index=False)

valid_rows = preprocessing_manifest.query("status == 'valid'").copy()
rejected_rows = preprocessing_manifest.query("status != 'valid'").copy()

assert valid_rows['crop_path'].map(lambda crop_path: Path(crop_path).is_file()).all(), 'Há crops válidos ausentes.'
print(f'Imagens cruas: {len(preprocessing_manifest):,}')
print(f'Crops válidos: {len(valid_rows):,}')
print(f'Rejeitadas: {len(rejected_rows):,}')
print(f'Manifesto de pré-processamento: {manifest_path}')

if not rejected_rows.empty:
    display(rejected_rows.groupby(['class_name', 'source', 'status']).size().rename('images').to_frame())

frame = valid_rows.rename(columns={'crop_path': 'path'})[['path', 'label', 'class_name', 'source', 'source_path']].reset_index(drop=True)

display(frame.groupby(['class_name', 'source']).size().rename('images').to_frame())
```

## Auditoria antes da divisão

A auditoria identifica arquivos binariamente idênticos por SHA-256. A verificação adicional de similaridade e de origem/identidade entre partições aparece após a divisão dos dados. Hashes perceptuais apenas sugerem candidatos à revisão; não identificam pessoas.

```python
def sha256_file(path: str) -> str:
    digest = hashlib.sha256()
    with open(path, 'rb') as file:
        for block in iter(lambda: file.read(1024 * 1024), b''):
            digest.update(block)
    return digest.hexdigest()

dataset_audit = frame.assign(sha256=frame['path'].map(sha256_file))
duplicates = dataset_audit.duplicated('sha256', keep=False)

print(f'Duplicatas binárias: {duplicates.sum():,}')
assert not duplicates.any(), 'Remova ou agrupe duplicatas antes de treinar.'

AUDIT_DIR = MODELS_DIR / 'audit'
AUDIT_DIR.mkdir(parents=True, exist_ok=True)

dataset_manifest_path = AUDIT_DIR / 'resnet50_224_dataset_manifest.csv'
dataset_audit.sort_values('path').to_csv(dataset_manifest_path, index=False)

print(f'Manifesto salvo em: {dataset_manifest_path}')
```

## Cenários de divisão

`random_image` divide imagens aleatoriamente. O experimento utiliza `source_holdout`: imagens fake da fonte `sd_21` e imagens reais da fonte `celebhq` ficam exclusivamente no teste. Esse protocolo avalia o par reservado, sem demonstrar generalização universal nem separar integralmente os efeitos de fonte e classe.

```python
def make_random_split(data: pd.DataFrame):
    train, temporary = train_test_split(
        data, test_size=0.40, stratify=data['label'], random_state=SEED,
    )
    validation, test = train_test_split(
        temporary, test_size=0.50, stratify=temporary['label'], random_state=SEED,
    )
    return train.copy(), validation.copy(), test.copy()


def make_source_holdout_split(data: pd.DataFrame, fake_test_source: str, real_test_source: str):
    test_mask = ((data.class_name == 'fake') & (data.source == fake_test_source)) | (
        (data.class_name == 'real') & (data.source == real_test_source)
    )
    test = data.loc[test_mask].copy()
    remaining = data.loc[~test_mask].copy()
    assert set(test.class_name) == set(CLASS_NAMES), 'O teste precisa conter ambas as classes.'
    train, validation = train_test_split(
        remaining, test_size=0.20, stratify=remaining['label'], random_state=SEED,
    )
    return train.copy(), validation.copy(), test

SCENARIO = 'source_holdout'
HOLDOUT_FAKE_SOURCE = 'sd_21'
HOLDOUT_REAL_SOURCE = 'celebhq'

if SCENARIO == 'random_image':
    train_df, validation_df, test_df = make_random_split(frame)
elif SCENARIO == 'source_holdout':
    train_df, validation_df, test_df = make_source_holdout_split(
        frame, HOLDOUT_FAKE_SOURCE, HOLDOUT_REAL_SOURCE,
    )
else:
    raise ValueError('SCENARIO inválido.')

for name, split in {'train': train_df, 'validation': validation_df, 'test': test_df}.items():
    print(f'\n{name}: {len(split):,}')
    display(split.groupby(['class_name', 'source']).size().rename('images').to_frame())

# Invariantes de isolamento: nenhum caminho pode existir em duas partições.
train_paths, val_paths, test_paths = map(set, (train_df.path, validation_df.path, test_df.path))
assert not train_paths & val_paths and not train_paths & test_paths and not val_paths & test_paths
if SCENARIO == 'source_holdout':
    assert HOLDOUT_FAKE_SOURCE not in set(train_df.query("class_name == 'fake'").source)
    assert HOLDOUT_REAL_SOURCE not in set(train_df.query("class_name == 'real'").source)

# A associação exata imagem→partição é parte do experimento e deve ser preservada.
split_manifest = pd.concat([
    train_df.assign(split='train'),
    validation_df.assign(split='validation'),
    test_df.assign(split='test'),
]).sort_values('path')

split_manifest_path = AUDIT_DIR / f'resnet50_224_{SCENARIO}_seed{SEED}_splits.csv'
split_manifest.to_csv(split_manifest_path, index=False)
print(f'Manifesto de partições: {split_manifest_path}')
```

## Auditoria complementar de semelhança, origem e partições

O hash perceptual **dHash** identifica *candidatos* a quase duplicatas entre os recortes, com distância de Hamming de até 5 bits. Pares em partições diferentes requerem revisão: a similaridade não comprova identidade nem vazamento por si só. O procedimento não substitui inspeção de identidade ou de imagem de origem.

Também são exportadas as contagens por fonte e partição e uma planilha de linhagem. Os identificadores `original_group_id` e `identity_id` devem ser preenchidos **somente se houver metadados confiáveis das fontes**; campos vazios não comprovam ausência de dependência. Origem, versão, URL e condições de uso precisam ser verificadas e preenchidas pelo grupo. Se forem identificados grupos compartilhados entre partições, a separação precisa ser revista antes de utilizar os resultados como independentes.

```python
# Totais por fonte, incluindo exclusões e composição de cada partição.
source_counts = preprocessing_manifest.groupby(['class_name', 'source']).size().rename('raw_images').to_frame()
source_counts['valid_images'] = valid_rows.groupby(['class_name', 'source']).size()
source_counts = source_counts.fillna(0).astype(int)
source_counts['rejected_images'] = source_counts['raw_images'] - source_counts['valid_images']
for name, subset in [('train', train_df), ('validation', validation_df), ('test', test_df)]:
    source_counts[name] = subset.groupby(['class_name', 'source']).size()
source_counts = source_counts.fillna(0).astype(int).reset_index()
for column in ['dataset_origin', 'dataset_version', 'source_url', 'license_or_terms']:
    source_counts[column] = ''  # Preencher somente com informações verificadas das fontes.
source_counts_path = AUDIT_DIR / 'resnet50_224_source_provenance.csv'
if source_counts_path.exists():
    previous = pd.read_csv(source_counts_path, dtype=str).fillna('')
    if {'class_name', 'source', 'dataset_origin', 'dataset_version', 'source_url', 'license_or_terms'} <= set(previous.columns):
        source_counts = source_counts.drop(columns=['dataset_origin', 'dataset_version', 'source_url', 'license_or_terms']).merge(
            previous[['class_name', 'source', 'dataset_origin', 'dataset_version', 'source_url', 'license_or_terms']],
            on=['class_name', 'source'], how='left', validate='one_to_one',
        ).fillna('')
source_counts.to_csv(source_counts_path, index=False)
print(f'Fontes, rejeições e partições: {source_counts_path}')

# Identificadores de pessoa/imagem original dependem de metadados externos confiáveis.
lineage_path = AUDIT_DIR / 'resnet50_224_identity_origin_review.csv'
if lineage_path.exists():
    lineage_review = pd.read_csv(lineage_path, dtype=str).fillna('')
    current_splits = split_manifest[['path', 'split']].sort_values('path').reset_index(drop=True)
    saved_splits = lineage_review[['path', 'split']].sort_values('path').reset_index(drop=True)
    if not current_splits.equals(saved_splits):
        raise RuntimeError('A planilha de identidade/origem pertence a outra divisão dos dados. Revise-a antes de continuar.')
else:
    lineage_review = split_manifest[['path', 'source_path', 'class_name', 'source', 'split']].copy()
    lineage_review['original_group_id'] = ''
    lineage_review['identity_id'] = ''
    lineage_review.to_csv(lineage_path, index=False)
print(f'Planilha de revisão de identidade/origem: {lineage_path}')

for id_column in ['original_group_id', 'identity_id']:
    identified = lineage_review.loc[lineage_review[id_column].ne('')]
    overlapping = identified.groupby(id_column)['split'].nunique()
    conflicting_groups = overlapping[overlapping > 1]
    print(f'Grupos de {id_column} identificados: {len(overlapping)}; presentes em partições distintas: {len(conflicting_groups)}')
    if len(conflicting_groups):
        raise RuntimeError(f'Existem grupos compartilhados entre partições em {id_column}. Corrija a divisão antes de treinar.')

# dHash de 64 bits: detecta candidatos visivelmente semelhantes, sem inferir identidade.
def dhash_file(path: str) -> int:
    image = cv2.imread(str(path), cv2.IMREAD_GRAYSCALE)
    if image is None:
        raise RuntimeError(f'Não foi possível ler o crop para auditoria perceptual: {path}')
    small = cv2.resize(image, (9, 8), interpolation=cv2.INTER_AREA)
    return int.from_bytes(np.packbits(small[:, 1:] > small[:, :-1]).tobytes(), 'big')

perceptual_frame = split_manifest[['path', 'class_name', 'source', 'split']].reset_index(drop=True)
perceptual_hashes = [dhash_file(path) for path in tqdm(perceptual_frame.path, desc='Auditoria dHash', unit='imagem')]

# Se dois hashes diferem em até 5 bits, pelo menos uma das seis faixas coincide.
bands = [(0, 11), (11, 11), (22, 11), (33, 11), (44, 10), (54, 10)]
buckets = defaultdict(list)
near_duplicate_pairs = {}
for index, current_hash in enumerate(perceptual_hashes):
    for band_id, (offset, width) in enumerate(bands):
        band_value = (current_hash >> (64 - offset - width)) & ((1 << width) - 1)
        key = (band_id, band_value)
        for other_index in buckets[key]:
            distance = (current_hash ^ perceptual_hashes[other_index]).bit_count()
            if distance <= 5:
                near_duplicate_pairs[(other_index, index)] = distance
        buckets[key].append(index)

candidate_rows = []
for (left, right), distance in near_duplicate_pairs.items():
    a, b = perceptual_frame.iloc[left], perceptual_frame.iloc[right]
    candidate_rows.append({
        'path_a': a.path, 'path_b': b.path,
        'class_a': a.class_name, 'class_b': b.class_name,
        'source_a': a.source, 'source_b': b.source,
        'split_a': a.split, 'split_b': b.split,
        'hamming_distance': distance, 'cross_split': a.split != b.split,
    })
near_duplicates = pd.DataFrame(candidate_rows, columns=[
    'path_a', 'path_b', 'class_a', 'class_b', 'source_a', 'source_b',
    'split_a', 'split_b', 'hamming_distance', 'cross_split',
]).sort_values('hamming_distance', ignore_index=True)
near_duplicates_path = AUDIT_DIR / 'resnet50_224_near_duplicate_candidates.csv'
near_duplicates.to_csv(near_duplicates_path, index=False)

cross_split_candidates = near_duplicates.loc[near_duplicates['cross_split']]
print(f'Candidatos dHash (<= 5 bits): {len(near_duplicates):,}')
print(f'Candidatos entre partições diferentes, sujeitos a revisão: {len(cross_split_candidates):,}')
print(f'Relatório de candidatos: {near_duplicates_path}')
if not cross_split_candidates.empty:
    display(cross_split_candidates.head(20))
```

## Preparação dos dados para treinamento e avaliação

Definição do Dataset personalizado, das transformações de aumento de dados (*data augmentation*) e da normalização das imagens.

As transformações aleatórias são aplicadas exclusivamente ao treinamento. Validação e teste utilizam apenas a conversão para tensor e a normalização.

Os DataLoaders são configurados com amostragem balanceada por fonte no treinamento, enquanto validação e teste mantêm a ordem dos registros.

```python
if "FaceDataset" not in globals():
    class FaceDataset(Dataset):
        def __init__(self, records: list[dict], transform):
            self.records = records
            self.transform = transform

        def __len__(self):
            return len(self.records)

        def __getitem__(self, index):
            record = self.records[index]
            with Image.open(record['path']) as image:
                image = image.convert('RGB')
            return self.transform(image), record['label']

class RandomJPEGCompression:
    """Simula recompressão comum de upload/redes sociais sem tocar validação/teste."""
    def __init__(self, probability=0.25, quality_range=(60, 95)):
        self.probability = probability
        self.quality_range = quality_range

    def __call__(self, image):
        if random.random() >= self.probability:
            return image
        quality = random.randint(*self.quality_range)
        buffer = io.BytesIO()
        image.save(buffer, format='JPEG', quality=quality)
        buffer.seek(0)
        with Image.open(buffer) as recompressed:
            return recompressed.convert('RGB')

normalize = transforms.Normalize(mean=MEAN, std=STD)
train_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomAffine(degrees=3, translate=(0.03, 0.03), scale=(0.97, 1.03)),
    transforms.RandomApply([transforms.ColorJitter(0.06, 0.06, 0.04, 0.01)], p=0.35),
    RandomJPEGCompression(JPEG_AUGMENTATION_PROBABILITY),
    transforms.RandomApply([transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 0.7))], p=0.10),
    transforms.ToTensor(), normalize,
])
eval_transform = transforms.Compose([transforms.ToTensor(), normalize])

train_records = train_df.to_dict('records')
if USE_SOURCE_BALANCED_SAMPLER:
    # Balanceia as fontes dentro de cada classe por meio de pesos de amostragem.
    class_counts = train_df.groupby('class_name').size().to_dict()
    sources_per_class = train_df.groupby('class_name')['source'].nunique().to_dict()
    group_counts = train_df.groupby(['class_name', 'source']).size().to_dict()
    sample_weights = [
        1.0 / (
            class_counts[row['class_name']]
            * sources_per_class[row['class_name']]
            * group_counts[(row['class_name'], row['source'])]
        )
        for row in train_records
    ]
    train_sampler = WeightedRandomSampler(
        weights=torch.as_tensor(sample_weights, dtype=torch.double),
        num_samples=len(train_records),
        replacement=True,
        generator=torch.Generator().manual_seed(SEED),
    )
    train_loader = DataLoader(FaceDataset(train_records, train_transform), BATCH_SIZE,
                              sampler=train_sampler, shuffle=False, num_workers=NUM_WORKERS, pin_memory=DEVICE.type == 'cuda')
else:
    train_loader = DataLoader(FaceDataset(train_records, train_transform), BATCH_SIZE,
                              shuffle=True, num_workers=NUM_WORKERS, pin_memory=DEVICE.type == 'cuda')

validation_loader = DataLoader(FaceDataset(validation_df.to_dict('records'), eval_transform), BATCH_SIZE,
                               shuffle=False, num_workers=NUM_WORKERS, pin_memory=DEVICE.type == 'cuda')

test_loader = DataLoader(FaceDataset(test_df.to_dict('records'), eval_transform), BATCH_SIZE,
                         shuffle=False, num_workers=NUM_WORKERS, pin_memory=DEVICE.type == 'cuda')

print(f'Sampler balanceado por fonte: {USE_SOURCE_BALANCED_SAMPLER}')
```

## ResNet50 e fine-tuning com regularização de domínio

A fase inicial adapta somente a cabeça; depois toda a rede é ajustada com learning rate menor no backbone. O treino aplica pesos de amostragem por classe e fonte, sem garantia de 50% exatos de cada classe. Augmentations leves simulam variações de câmera e compressão sem destruir completamente os artefatos forenses.

```python
def build_model():
    model = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1)
    model.fc = nn.Linear(model.fc.in_features, len(CLASS_NAMES))
    return model.to(DEVICE)

model = build_model()
criterion = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTHING)

for parameter in model.parameters():
    parameter.requires_grad = False
for parameter in model.fc.parameters():
    parameter.requires_grad = True

optimizer = optim.AdamW(model.fc.parameters(), lr=LR_HEAD, weight_decay=WEIGHT_DECAY)
print('Fase 1: somente a camada final treinável.')
```

## Treinamento da ResNet50 e monitoramento da validação

O treinamento é realizado em duas fases: inicialmente, apenas a camada de classificação é treinada, mantendo o backbone pré-treinado congelado. Em seguida, todas as camadas são liberadas para ajuste fino (*fine-tuning*).

Durante o treinamento, são registrados a perda de treinamento, a perda de validação, a acurácia e a ROC-AUC de validação.

São utilizados escalonamento adaptativo da taxa de aprendizado e parada antecipada (*early stopping*). Ao final, são restaurados os pesos correspondentes à menor perda de validação.

A execução também registra o ambiente computacional, as versões das bibliotecas, a duração observada e a época escolhida para o checkpoint.

```python
@torch.inference_mode()
def predict_loader(loader):
    model.eval()
    targets, probabilities = [], []
    for images, labels in loader:
        logits = model(images.to(DEVICE, non_blocking=True))
        probabilities.extend(torch.softmax(logits, dim=1)[:, CLASS_TO_INDEX['fake']].cpu().numpy())
        targets.extend(labels.numpy())
    return np.asarray(targets), np.asarray(probabilities)


def one_epoch(loader, backbone_frozen: bool = False):
    model.train()
    # Com backbone congelado, BatchNorm também deve permanecer em avaliação;
    # caso contrário suas estatísticas seriam alteradas mesmo sem gradientes.
    if backbone_frozen:
        for module in model.modules():
            if isinstance(module, nn.modules.batchnorm._BatchNorm):
                module.eval()
    total_loss, total = 0.0, 0
    for images, labels in loader:
        images, labels = images.to(DEVICE, non_blocking=True), labels.to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        loss = criterion(model(images), labels)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRADIENT_CLIP_NORM)
        optimizer.step()
        total_loss += loss.item() * labels.size(0)
        total += labels.size(0)
    return total_loss / total

training_started_at = datetime.now(timezone.utc).isoformat()
training_start = time.perf_counter()
history = []
best_epoch = None
best_loss = float('inf')
best_state = None
stale_epochs = 0
scheduler = None

for epoch in range(1, EPOCHS + 1):
    if epoch == HEAD_EPOCHS + 1:
        for parameter in model.parameters():
            parameter.requires_grad = True
        backbone_parameters = [
            parameter for name, parameter in model.named_parameters()
            if not name.startswith('fc.')
        ]
        optimizer = optim.AdamW([
            {'params': backbone_parameters, 'lr': LR_BACKBONE},
            {'params': model.fc.parameters(), 'lr': LR_HEAD},
        ], weight_decay=WEIGHT_DECAY)
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.3, patience=2)
        stale_epochs = 0
        print('Fase 2: fine-tuning completo iniciado.')

    train_loss = one_epoch(train_loader, backbone_frozen=epoch <= HEAD_EPOCHS)
    validation_targets, validation_probabilities = predict_loader(validation_loader)
    model.eval()
    ce_sum, ce_count = 0.0, 0
    with torch.inference_mode():
        for images, labels in validation_loader:
            loss = criterion(model(images.to(DEVICE)), labels.to(DEVICE))
            ce_sum += loss.item() * labels.size(0); ce_count += labels.size(0)
    validation_loss = ce_sum / ce_count
    validation_accuracy = 100 * ((validation_probabilities >= 0.5) == (validation_targets == CLASS_TO_INDEX['fake'])).mean()
    validation_auc = roc_auc_score(validation_targets == CLASS_TO_INDEX['fake'], validation_probabilities)
    history.append({'epoch': epoch, 'train_loss': train_loss, 'validation_loss': validation_loss,
                    'validation_accuracy': validation_accuracy, 'validation_auc': validation_auc})
    print(f'Epoch {epoch:02d} | train loss {train_loss:.4f} | val loss {validation_loss:.4f} | val acc {validation_accuracy:.2f}% | val AUC {validation_auc:.4f}')

    if scheduler is not None:
        scheduler.step(validation_loss)
    if validation_loss < best_loss - 1e-4:
        best_loss, best_state, stale_epochs = validation_loss, copy.deepcopy(model.state_dict()), 0
        best_epoch = epoch
    else:
        stale_epochs += 1
        if epoch > HEAD_EPOCHS and stale_epochs >= PATIENCE:
            print('Early stopping.'); break

training_duration_seconds = time.perf_counter() - training_start
assert best_state is not None
model.load_state_dict(best_state)
history = pd.DataFrame(history)
display(history)

history_path = AUDIT_DIR / 'resnet50_224_training_history.csv'
history.to_csv(history_path, index=False)
print(f'Histórico do treinamento salvo em: {history_path}')

# Ambiente e duração observados nesta execução (não inferidos de saídas antigas).
try:
    ram_bytes = os.sysconf('SC_PAGE_SIZE') * os.sysconf('SC_PHYS_PAGES')
except (AttributeError, ValueError, OSError):
    ram_bytes = None

gpu_properties = torch.cuda.get_device_properties(0) if DEVICE.type == 'cuda' else None
training_environment = {
    'started_at_utc': training_started_at,
    'duration_seconds': round(training_duration_seconds, 2),
    'epochs_executed': len(history),
    'best_epoch': best_epoch,
    'checkpoint_selection': 'menor perda de validação',
    'system': platform.platform(),
    'cpu': platform.processor() or 'não identificado',
    'cpu_logical_count': os.cpu_count(),
    'ram_gb': round(ram_bytes / (1024 ** 3), 2) if ram_bytes is not None else None,
    'gpu': torch.cuda.get_device_name(0) if gpu_properties is not None else None,
    'gpu_memory_gb': round(gpu_properties.total_memory / (1024 ** 3), 2) if gpu_properties is not None else None,
    'python': sys.version.split()[0],
    'pytorch': str(torch.__version__),
    'torchvision': version('torchvision'),
    'opencv': cv2.__version__,
    'pillow': version('Pillow'),
    'scikit_learn': version('scikit-learn'),
    'seed': SEED,
    'image_size': IMAGE_SIZE,
    'batch_size': BATCH_SIZE,
    'num_workers': NUM_WORKERS,
    'max_epochs': EPOCHS,
    'head_epochs': HEAD_EPOCHS,
    'early_stopping_patience': PATIENCE,
    'lr_head': LR_HEAD,
    'lr_backbone': LR_BACKBONE,
    'weight_decay': WEIGHT_DECAY,
    'label_smoothing': LABEL_SMOOTHING,
    'gradient_clip_norm': GRADIENT_CLIP_NORM,
    'source_balanced_sampler': USE_SOURCE_BALANCED_SAMPLER,
    'jpeg_augmentation_probability': JPEG_AUGMENTATION_PROBABILITY,
    'yunet_score_threshold': FACE_DETECTOR_SCORE_THRESHOLD,
    'face_margin': FACE_MARGIN,
    'normalization': {'mean': MEAN, 'std': STD},
}
environment_path = AUDIT_DIR / 'resnet50_224_training_environment.json'
with open(environment_path, 'w', encoding='utf-8') as file:
    json.dump(training_environment, file, indent=2, ensure_ascii=False)
print(f'Ambiente e duração do treinamento: {environment_path}')
```

## Seleção do limiar e avaliação final do modelo

O limiar de decisão é selecionado exclusivamente no conjunto de validação, por maximização da métrica F2 para a classe `fake`, priorizando a identificação de imagens sintéticas.

Após a definição do limiar, o modelo é avaliado no conjunto de teste reservado, utilizando métricas de classificação, ROC-AUC, Average Precision (AP) e matrizes de confusão absoluta e normalizada.

Também é realizada uma análise dos resultados por classe e fonte de origem, permitindo examinar o desempenho no cenário `source_holdout`.

O empate em F2 é resolvido pelo menor limiar da grade crescente (primeiro máximo de `np.argmax`). A saída `softmax` é tratada como escore, não como probabilidade calibrada; com limiar abaixo de 0,50, a classe selecionada pode não ter o maior escore. A AP é calculada por `average_precision_score`, não por integração trapezoidal.

```python
validation_targets, validation_probabilities = predict_loader(validation_loader)

thresholds = np.linspace(0.01, 0.99, 99)
validation_f2 = [
    fbeta_score(validation_targets == CLASS_TO_INDEX['fake'], validation_probabilities >= t, beta=2)
    for t in thresholds
]
validation_precision = [
    precision_score(validation_targets == CLASS_TO_INDEX['fake'], validation_probabilities >= t, zero_division=0)
    for t in thresholds
]
validation_recall = [
    recall_score(validation_targets == CLASS_TO_INDEX['fake'], validation_probabilities >= t, zero_division=0)
    for t in thresholds
]
# np.argmax retorna o primeiro máximo; a grade crescente desempata pelo menor limiar.
threshold = float(thresholds[int(np.argmax(validation_f2))])

# Registra os resultados da seleção do limiar na validação.
threshold_report = pd.DataFrame({
    'threshold': thresholds, 'precision_fake': validation_precision,
    'recall_fake': validation_recall, 'f2_fake': validation_f2,
})
threshold_report_path = AUDIT_DIR / 'resnet50_224_threshold_selection.csv'
threshold_report.to_csv(threshold_report_path, index=False)

print(f'Limiar selecionado na validação: escore softmax fake >= {threshold:.2f}')
print(f'F2 máximo na validação: {max(validation_f2):.4f}')
print(f'Auditoria do limiar: {threshold_report_path}')

plt.figure(figsize=(8, 4))
plt.plot(thresholds, validation_f2)
plt.axvline(threshold, linestyle='--', label=f'Limiar selecionado: {threshold:.2f}')
plt.xlabel('Limiar de decisão')
plt.ylabel('F2 da classe fake')
plt.title('Seleção do limiar no conjunto de validação')
plt.legend()
plt.tight_layout()
plt.savefig(AUDIT_DIR / 'resnet50_224_threshold_f2.png', dpi=200, bbox_inches='tight')
plt.show()

# O teste é utilizado somente após a seleção do limiar.
test_targets, test_probabilities = predict_loader(test_loader)
test_predictions = np.where(
    test_probabilities >= threshold,
    CLASS_TO_INDEX['fake'],
    CLASS_TO_INDEX['real']
)

print(f'ROC-AUC teste: {roc_auc_score(test_targets == CLASS_TO_INDEX["fake"], test_probabilities):.4f}')
print(f'Average Precision (AP) teste: {average_precision_score(test_targets == CLASS_TO_INDEX["fake"], test_probabilities):.4f}')
print(classification_report(test_targets, test_predictions, target_names=CLASS_NAMES, digits=4))
print('Matriz de confusão (linhas: classe real; colunas: classe prevista):')
print(confusion_matrix(test_targets, test_predictions, labels=[CLASS_TO_INDEX['fake'], CLASS_TO_INDEX['real']]))
```

```python
labels_order = [CLASS_TO_INDEX['fake'], CLASS_TO_INDEX['real']]
display_labels = ['fake', 'real']

cm_counts = confusion_matrix(test_targets, test_predictions, labels=labels_order)
cm_normalized = confusion_matrix(
    test_targets, test_predictions, labels=labels_order, normalize='true',
)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for axis, matrix, title, fmt in [
    (axes[0], cm_counts, 'Matriz de confusão — contagem', 'd'),
    (axes[1], cm_normalized, 'Matriz de confusão — normalizada por classe', '.2%'),
]:
    sns.heatmap(
        matrix, annot=True, fmt=fmt, cmap='Blues', cbar=False,
        xticklabels=display_labels, yticklabels=display_labels, ax=axis,
        vmin=0 if fmt == '.2%' else None, vmax=1 if fmt == '.2%' else None,
    )
    axis.set_xlabel('Classe prevista')
    axis.set_ylabel('Classe real')
    axis.set_title(title)

plt.tight_layout()
plt.savefig(AUDIT_DIR / 'resnet50_224_confusion_matrices.png', dpi=200, bbox_inches='tight')
plt.show()

# Curvas ROC e Precisão-Recall no conjunto de teste.
binary_targets = test_targets == CLASS_TO_INDEX['fake']
fpr, tpr, _ = roc_curve(binary_targets, test_probabilities)
precision, recall, _ = precision_recall_curve(binary_targets, test_probabilities)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(fpr, tpr)
axes[0].plot([0, 1], [0, 1], linestyle='--')
axes[0].set_xlabel('Taxa de falsos positivos')
axes[0].set_ylabel('Taxa de verdadeiros positivos')
axes[0].set_title('Curva ROC — teste')

axes[1].plot(recall, precision)
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precisão')
axes[1].set_title('Curva Precisão-Recall — teste')

plt.tight_layout()
plt.savefig(AUDIT_DIR / 'resnet50_224_roc_pr_curves.png', dpi=200, bbox_inches='tight')
plt.show()

# Resultados individuais e métricas por fonte no cenário source_holdout.
test_report = test_df[['path', 'class_name', 'source']].copy().reset_index(drop=True)
test_report['probability_fake'] = test_probabilities
test_report['prediction'] = [CLASS_NAMES[index] for index in test_predictions]
test_report['correct'] = test_report['prediction'].eq(test_report['class_name'])

source_summary = (
    test_report.groupby(['class_name', 'source'])
    .agg(
        images=('correct', 'size'),
        accuracy=('correct', 'mean'),
        probability_fake_mean=('probability_fake', 'mean'),
    )
    .sort_index()
)

display(
    source_summary.style.format({
        'accuracy': '{:.2%}',
        'probability_fake_mean': '{:.3f}',
    })
)

test_report_path = AUDIT_DIR / 'resnet50_224_test_predictions.csv'
source_summary_path = AUDIT_DIR / 'resnet50_224_source_summary.csv'

test_report.to_csv(test_report_path, index=False)
source_summary.to_csv(source_summary_path)

print(f'Predições individuais: {test_report_path}')
print(f'Resumo por fonte: {source_summary_path}')
```
